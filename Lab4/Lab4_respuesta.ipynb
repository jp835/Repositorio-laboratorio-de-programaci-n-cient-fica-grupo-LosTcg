```python
"""
En la siguiente celda se crean los datos ficticios a usar en la sección 1 del lab.
❗No realice cambios a esta celda a excepción de n_samples❗
"""

# Datos a utilizar

# Configuracion
n_samples = 5000 #Este parámetro si lo pueden modificar

def create_data(n_samples):

    # Lunas
    moons = datasets.make_moons(n_samples=n_samples, noise=0.05, random_state=30)
    # Blobs
    blobs = datasets.make_blobs(n_samples=n_samples, random_state=172)
    # Datos desiguales
    transformation = [[0.6, -0.6], [-0.4, 0.8]]
    mutated = (np.dot(blobs[0], transformation), blobs[1])

    # Generamos Dataset
    dataset = {
        'moons':{
            'x': moons[0], 'classes': moons[1], 'n_cluster': 2
        },
        'blobs':{
            'x': blobs[0], 'classes': blobs[1], 'n_cluster': 3
        },
        'mutated':{
            'x': mutated[0], 'classes': mutated[1], 'n_cluster': 3
        }
    }
    return dataset

data_sets = create_data(n_samples)
```

```python
import time
import numpy as np
import plotly.subplots as sp
import plotly.graph_objects as go

from sklearn import datasets
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score
```

**Respuestas:**

```python
# Función para entrenar clustering

def fit_cluster(X, n_clusters, method="kmeans"):
    start = time.time()

    if method == "kmeans":
        model = KMeans(n_clusters=n_clusters, n_init=10, random_state=42)
        labels = model.fit_predict(X)

    elif method == "gmm":
        model = GaussianMixture(n_components=n_clusters, random_state=42)
        labels = model.fit_predict(X)

    elif method == "ward":
        model = AgglomerativeClustering(n_clusters=n_clusters, linkage='ward')
        labels = model.fit_predict(X)

    elif method == "dbscan":
        model = DBSCAN(eps=0.2, min_samples=3)
        labels = model.fit_predict(X)

    else:
        raise ValueError("Método no reconocido")

    end = time.time()
    elapsed = end - start

    # Silhouette solo si hay más de 1 cluster
    if len(set(labels)) > 1 and -1 not in labels:
        score = silhouette_score(X, labels)
    else:
        score = np.nan

    return labels, elapsed, score


# Función para la figura

def comparar_metodos(n_samples):
    data_sets = create_data(n_samples)

    methods = ["kmeans", "gmm", "ward", "dbscan"]
    titles = ["KMeans", "GMM", "WARD", "DBSCAN"]

    fig = sp.make_subplots(rows=3, cols=4,
                           subplot_titles=[t for _ in range(3) for t in titles],
                           horizontal_spacing=0.05,
                           vertical_spacing=0.1)

    row_map = {"moons": 1, "blobs": 2, "mutated": 3}

    for j, method in enumerate(methods, start=1):
        for i, (name, dset) in enumerate(data_sets.items(), start=1):
            X, n_clusters = dset['x'], dset['n_cluster']

            labels, t, s = fit_cluster(X, n_clusters, method)

            scatter = go.Scatter(
                x=X[:, 0], y=X[:, 1],
                mode='markers',
                marker=dict(color=labels, colorscale='Viridis', size=4),
                showlegend=False
            )

            r = row_map[name]
            fig.add_trace(scatter, row=r, col=j)

            # Texto con tiempo y score
            fig.add_annotation(
                text=f"{t:.2f} s | s: {s:.2f}",
                xref=f"x{(r-1)*4+j}", yref=f"y{(r-1)*4+j}",
                x=0.5, y=1.1, showarrow=False,
                xanchor="center", yanchor="bottom",
                font=dict(size=10)
            )

    fig.update_layout(
        height=900, width=1200,
        title_text=f"Comparación de tiempos de ejecución por técnica (n_samples={n_samples})",
        title_x=0.5
    )
    return fig


# Ejecutar para distintos tamaños

for n in [1000, 5000, 10000]:
    fig = comparar_metodos(n)
    fig.show()
```

```python
df_aerolinea = pd.read_parquet('aerolineas_lucer.parquet')
df_aero_numericas_id = df_aerolinea.select_dtypes(include='number')
df_aero_numericas = df_aero_numericas_id.drop(columns = ['id'])


n_vars = len(df_aero_numericas.columns)
cols = 3
rows = (n_vars + cols - 1) // cols  # ceil división entera

# Creamos figura con subplots
fig = make_subplots(rows=rows, cols=cols, subplot_titles=df_aero_numericas.columns)

# Agregamos un histograma por variable
for i, col_name in enumerate(df_aero_numericas.columns):
    row = i // cols + 1
    col = i % cols + 1

    fig.add_trace(
        go.Histogram(x=df_aero_numericas[col_name], marker_color='teal', opacity=0.75),
        row=row,
        col=col
    )

# Ajustes de layout
fig.update_layout(height=rows*300, width=1000, showlegend=False, title_text="Histogramas variables numéricas escaladas")

fig.show()
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df_numericas_scaled = pd.DataFrame(scaler.fit_transform(df_aero_numericas), columns=df_aero_numericas.columns)




n_vars = len(df_numericas_scaled.columns)
cols = 3
rows = (n_vars + cols - 1) // cols  # ceil división entera

# Creamos figura con subplots
fig = make_subplots(rows=rows, cols=cols, subplot_titles=df_numericas_scaled.columns)

# Agregamos un histograma por variable
for i, col_name in enumerate(df_numericas_scaled.columns):
    row = i // cols + 1
    col = i % cols + 1

    fig.add_trace(
        go.Histogram(x=df_numericas_scaled[col_name], marker_color='teal', opacity=0.75),
        row=row,
        col=col
    )

# Ajustes de layout
fig.update_layout(height=rows*300, width=1000, showlegend=False, title_text="Histogramas variables numéricas escaladas")

fig.show()
```

```python
#corr_matrix = df_aero_numericas.corr()
corr_matrix = df_numericas_scaled.corr()
# Correlograma con plotly express
fig = px.imshow(
    corr_matrix,
    text_auto=True,  
    color_continuous_scale='RdBu_r',
    zmin=-1,
    zmax=1,
    title="Correlograma (Matriz de Correlación)"
)

fig.update_layout(width=800, height=800)
fig.show()
```

```python
remaining_vars = corr_matrix.columns.tolist()

# Iteramos hasta quedarnos con 4
while len(remaining_vars) > 4:
    # Submatriz actual
    sub_corr = corr_matrix.loc[remaining_vars, remaining_vars]

    # Ponemos NaN en la diagonal para ignorar correlación consigo misma
    np.fill_diagonal(sub_corr.values, np.nan)

    # Encontrar el par más correlacionado
    max_corr = sub_corr.unstack().idxmax()
    var1, var2 = max_corr

    # Calcular la suma de correlaciones de cada una con las demás
    sum_corr1 = sub_corr.loc[var1].sum()
    sum_corr2 = sub_corr.loc[var2].sum()

    # Eliminar la más "redundante" (mayor suma de correlaciones)
    to_drop = var1 if sum_corr1 > sum_corr2 else var2
    remaining_vars.remove(to_drop)

# Resultado final
print("Variables seleccionadas:", remaining_vars)
```

Al analizar los graficos realizados, en este caso se usaron los criterios AIC y BIC, los cuales optimizan los clusters en base a verosimilitud, podemos ver que en ambos caso el mejor valor es 7, ya que cuentan con el valor mas pequeño. Se escoje este en base a su formula cual es AIC=2k−2ln(L) y BIC=kln(N)−2ln(L), en el que se pueden ver distintas penalizaciones, una siendo 2k y el de la otra siendo kln(N), es decir el segundo penaliza mas a medida que hay mas datos y preferira modelos mas simple, en el otro caso, el AIC penaliza menos a bases mas grandes, por lo que se puede decir que es menos restrictivos. En este caso ambos dieron el mismo resultado, por lo que no hay discusion en cual es mas optimo, pero si fuesen a ser distintos se debateria la importancia de los errores y que tan restrictivo tendria que ser.

```python
from sklearn.mixture import GaussianMixture
#pipeline_gmm = Pipeline([
#    ('scaler', MinMaxScaler()),
#    ('gmm', GaussianMixture(n_components=3, random_state=42))
#])

X_selected = df_aero_numericas[selected_vars]
aic_scores = []
bic_scores = []

for n_clusters in range(3,9):
    pipeline_gmm = Pipeline([
        ('scaler', MinMaxScaler()),
        ('gmm', GaussianMixture(n_components=n_clusters, random_state=42))
    ])
    pipeline_gmm.fit(X_selected)
    X_scaled = pipeline_gmm.named_steps['scaler'].transform(X_selected)
    
    # Calculamos AIC y BIC usando el GMM dentro del pipeline
    gmm = pipeline_gmm.named_steps['gmm']
    aic_scores.append(gmm.aic(X_scaled))
    bic_scores.append(gmm.bic(X_scaled))



n_clusters = list(range(3, 9))

fig = make_subplots(rows=2, cols=1, shared_xaxes=True,
                    subplot_titles=("AIC vs Número de clusters", "BIC vs Número de clusters"))

# Gráfico AIC
fig.add_trace(go.Scatter(
    x=n_clusters,
    y=aic_scores,
    mode='lines+markers',
    name='AIC',
    line=dict(color='blue')
), row=1, col=1)

# Gráfico BIC
fig.add_trace(go.Scatter(
    x=n_clusters,
    y=bic_scores,
    mode='lines+markers',
    name='BIC',
    line=dict(color='red')
), row=2, col=1)


fig.update_layout(height=600, width=700,
                  title_text="Comparación de AIC y BIC para diferentes números de clusters",
                  template='plotly_white')

fig.update_xaxes(title_text="Número de clusters", row=2, col=1)
fig.update_yaxes(title_text="AIC", row=1, col=1)
fig.update_yaxes(title_text="BIC", row=2, col=1)

fig.show()
    
    

```
