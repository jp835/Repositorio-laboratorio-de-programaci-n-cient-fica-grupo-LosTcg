### 📌 Análisis individual [3.0 puntos]

````python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

clientes_df = pd.read_parquet('data/clientes.parquet')
productos_df = pd.read_parquet('data/productos.parquet')

universe_size = len(clientes_df) * len(productos_df)

def get_batch_performance(batch_id):
    # Cargar datos reales
    data = pd.read_parquet(f"data/batch_t{batch_id}.parquet")
    data = data.groupby(['customer_id', 'product_id']).size().reset_index()
    true_pairs = set(map(tuple, data[['customer_id', 'product_id']].values))

    # Cargar predicciones
    preds = pd.read_csv(f"predictions/batch_t{batch_id}.csv",
                        names=['customer_id', 'product_id'])
    pred_pairs = set(map(tuple, preds[['customer_id', 'product_id']].values))

    # TP = predicho y real
    tp = len(pred_pairs & true_pairs)
    # FP = predicho pero no real
    fp = len(pred_pairs - true_pairs)
    # FN = real pero no predicho
    fn = len(true_pairs - pred_pairs)
    # TN = universo total - todo lo demás
    tn = universe_size - tp - fp - fn

    y_true = [1]*tp + [0]*fp + [1]*fn + [0]*tn
    y_pred = [1]*tp + [1]*fp + [0]*fn + [0]*tn

    metrics = {
        "accuracy": accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred),
        "recall": recall_score(y_true, y_pred),
        "f1": f1_score(y_true, y_pred)
    }

    return metrics
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from IPython.display import display, Image, Markdown
# CARGAR DATOS DE FEATURE IMPORTANCE (Rutas Corregidas)

print("Cargando datos de importancia de variables...")

# Diccionario con las rutas exactas según tu imagen
rutas_features = {
    1: "Features_importances/feature_importances_fase_1.csv",
    2: "Features_importances/feature_importances_fase_2.csv",
    3: "Features_importances/feature_importances_fase_3.csv",
    4: "Features_importances/feature_importances_fase_4.csv"
}

dfs = []
for batch_id, ruta in rutas_features.items():
    if os.path.exists(ruta):
        # Cargar CSV
        df = pd.read_csv(ruta)
        df.columns = [c.lower() for c in df.columns]
        
        # Identificar columna de nombre y columna de valor
        col_feat = [c for c in df.columns if 'feature' in c or 'col' in c][0]
        col_imp = [c for c in df.columns if 'import' in c or 'gain' in c or 'weight' in c][0]
        
        # Renombrar y preparar
        df = df[[col_feat, col_imp]].rename(columns={col_feat: 'feature', col_imp: f'Batch {batch_id}'})
        df = df.set_index('feature')
        dfs.append(df)

# Consolidar todo en un solo DataFrame
df_feats_all = pd.concat(dfs, axis=1)



# Datos de métricas
data_metrics = {
    'Batch': ['Batch 1', 'Batch 2', 'Batch 3', 'Batch 4'],
    'Precision': [0.189, 0.271, 0.271, 0.269], 
    'Recall': [0.431, 0.395, 0.412, 0.415],     
    'F1 Score': [0.263, 0.321, 0.327, 0.326]    
}
df_metrics = pd.DataFrame(data_metrics)


# Gráficos
fig, axes = plt.subplots(2, 1, figsize=(12, 14))

# Gráfico 1
sns.lineplot(data=df_metrics, x='Batch', y='Precision', marker='o', label='Precision', ax=axes[0], linewidth=2)
sns.lineplot(data=df_metrics, x='Batch', y='Recall', marker='s', label='Recall', ax=axes[0], linewidth=2)
sns.lineplot(data=df_metrics, x='Batch', y='F1 Score', marker='^', label='F1 Score', ax=axes[0], linewidth=3, linestyle='--')

axes[0].set_title('Evolución del Performance del Modelo (Batches 1-4)', fontsize=14)
axes[0].set_ylabel('Score', fontsize=12)
axes[0].grid(True, linestyle='--', alpha=0.6)
axes[0].legend(fontsize=11)
axes[0].set_ylim(0.1, 0.5)

# Anotaciones
for i, txt in enumerate(df_metrics['F1 Score']):
    axes[0].annotate(f"{txt:.3f}", (i, df_metrics['F1 Score'][i]), 
                     textcoords="offset points", xytext=(0,10), ha='center', fontweight='bold')


df_feats_all['mean_imp'] = df_feats_all.mean(axis=1)
df_top_10 = df_feats_all.sort_values('mean_imp', ascending=False).head(10).drop(columns='mean_imp')


# Gráfico 2
sns.heatmap(df_top_10, annot=True, cmap='viridis', fmt=".3f", ax=axes[1])
axes[1].set_title('Top 10 Variables más importantes por Batch', fontsize=14)
axes[1].set_ylabel('Variable')


plt.tight_layout()
plt.show()

# Gráficos de Optimización (Optuna / SHAP)

display(Markdown("Optimización de Hiperparámetros (History Plot)"))
dir_imgs = "param importnaces"
imgs_disponibles = sorted([f for f in os.listdir(dir_imgs) if f.endswith('.png')]) if os.path.exists(dir_imgs) else []


for img_file in imgs_disponibles:
    display(Markdown(f"**{img_file}**"))
    display(Image(filename=os.path.join(dir_imgs, img_file), width=600))
````

3. Estrategia de detección: No se implementó un Data Drift como KS para activar el pipeline. En su lugar, asumimos un drift temporal; entonces, cada vez que llegaba un nuevo batch semanal se entrenaría otra semana. Trabajamos con la premisa de que el comportamiento de compra evoluciona semana a semana y que el modelo siempre mejorará con la información más reciente.

Criterio de cambio: Si la cantidad de archivos es superior a lo previo, el sistema asume que hay información relevante que debe ser incorporada.

Al analizar las métricas entre el batch 2 y el 4, en el que el F1 varió menos que 0.01, no se detectó un drift significativo. Si bien las distribuciones de entrada cambiaron, la relación entre estas variables y el objetivo se mantuvo igual.

4. Estrategia de reentrenamiento: Se diseñó una estrategia de entrenamiento que acumulaba periódicamente.

Lógica: El pipeline verifica periódicamente la cantidad de archivos.

Si aumenta la cantidad de archivos, se reentrena.

Esta decisión se tomó para mitigar el problema observado en el batch 1. Al no depender de un drift, garantizamos que el modelo siempre tenga todo el contexto histórico disponible.

Paso de entrenar solo con la base inicial (F1: 0.26) a entrenar con la data de dos batches, lo que elevó el F1 a 0.32.

5. Mayor impacto positivo: La decisión más valiosa fueron los lags y la recencia, como productos_diferentes y total_compras_producto. Al hacer atributos del comportamiento de las semanas $t-1$ a $t-4$, el modelo captura las tendencias del cliente. Esto se validó con el rendimiento del batch 1 al batch 2; estas variables aportaron la mayor ganancia de información.

Mayor impacto negativo: Debido al desbalance de clases, el modelo entregaba probabilidades que pocas veces superan 0.4. Al usar un umbral estándar ($>0.5$), el recall caía a 0.

6. Se utilizó el framework de Optuna para la búsqueda bayesiana de hiperparámetros, maximizando la métrica F1.

Hiperparámetros influyentes:

learning_rate: El parámetro más sensible; se seleccionaron valores bajos para que el gradient descent convergiera a la tendencia general.

scale_pos_weight: Dado el desbalance de clases, este parámetro fue importantísimo para penalizar más los errores en la clase compra, evitando que el modelo prediga siempre no compra.

max_depth: Reguló la complejidad del árbol.

7. ¿Qué variable fue más influyente en las predicciones?

Variable dominante: num__items_roll4_mean (promedio móvil de cantidad de ítems comprados en las últimas 4 semanas).

Interpretación: El modelo aprendió que el mejor predictor del futuro es el pasado del último mes. Si un cliente compra constantemente un producto durante el último mes, la probabilidad de que lo compre en la semana actual es mucho más alta.

8. ¿Qué aprendieron sobre el negocio a partir de los resultados del modelo?

Se confirmó que el negocio de SodAI funciona principalmente con reposición más que con nuevos clientes, con compras grandes. Los clientes son rutinarios y se puede obtener gran valor instalando un sistema de compra semanal habitual.

La consistencia de los batches 2, 3 y 4 sugiere que no hubo eventos estacionales disruptivos y que hubo una demanda consistente; hay que analizar posible estacionalidad.

El sistema actual es muy bueno para ahorrar tiempo al cliente al sugerir los productos que probablemente necesite. Sin embargo, es ineficiente para vender nuevos productos o conseguir stock de productos poco probables, ya que las variables de similitud no son relevantes.

9. ¿Qué limitaciones detectaron en su modelo o en los datos?

Hay un sesgo de historia: como en el batch 1, el modelo es mediocre debido a que no hay actividad reciente y luego se hace muy dependiente de la historia.

También hay una saturación de recomendaciones, ya que sin el filtro Top-5 el modelo carece de especificidad. Tiende a recomendar muchos productos y, sin la cota, es una recomendación muy vaga.

El modelo tampoco toma en cuenta precios, cambios de precios ni acciones de la competencia, por lo que las recomendaciones pueden perjudicar mucho a SodAI.
